```python
from bs4 import BeautifulSoup

with open('/content/job_description.txt', 'r') as file:
    html_content = file.read()

soup = BeautifulSoup(html_content, 'html.parser')
text_content = soup.get_text(separator=' ', strip=True)

# Print the text content, replacing '. ' with '.\n' for better readability
print(text_content)
```

```python
import io
import pandas as pd

# Treat the text_content as a file-like object
data_io = io.StringIO(text_content)

# Read the data using pandas, specifying the pipe separator
df = pd.read_csv(data_io, sep='|', index_col='id')

# Extract the 'title' and 'description' columns as Series
title = df['title']
description= df['description']

print("Title Series Head:")
display(title.head())

print("\nDescription Series Head:")
display(description.head())
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
import nltk
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
def text_clean(text):
  #text=text.lower()
  #converting to tokens
  tokens=word_tokenize(text)
  #stopwords removal
  stop_words=set(stopwords.words('english'))
  filterd_tokens=[word for word in tokens if word not in stop_words]
  #now that we have filtered tokens without stopwords
  stemmer=PorterStemmer()
  stemmed_tokens=[stemmer.stem(word) for word in filterd_tokens]
  cleaned_document=" ".join(stemmed_tokens)
  return cleaned_document
```

1.function text_cleaned built to tokenize,stemming and stopwrds.

2.Loaded job description.txt with the help of beaitiful soup.

3.Divided the txt file into title and description series
bold text

```python
#LETS CONVERT THE JOB DESCRIPTION TO NUMBERS USING TFIDF VECTORIZER
from sklearn.feature_extraction.text import TfidfVectorizer

cleaned_description=description.apply(text_clean)
#description series ma text_clean function apply gareko to clean
tfidf=TfidfVectorizer()
#class creation of tfidf vectorizer
tfidf_matrix=tfidf.fit_transform(cleaned_description)
#lets see the matrix
print(tfidf_matrix.shape)
```

```python
def process_text_to_vector(raw_text, fitted_tfidf):


    cleaned = text_clean(raw_text)


    vector = fitted_tfidf.transform([cleaned])

    return vector
```

```python
resume1="A motivated and detail-oriented Machine Learning Engineer with a strong foundation in Python, machine learning algorithms, data preprocessing, and model deployment. Experienced in developing predictive models, feature engineering, and evaluating model performance using libraries such as Scikit-learn, TensorFlow, and PyTorch. Passionate about solving real-world problems through artificial intelligence and continuously improving technical skills while collaborating effectively within cross-functional teams."
#resume2="A data-driven and analytical Data Scientist with expertise in statistical analysis, data visualization, and extracting actionable insights from complex datasets. Proficient in Python, SQL, Pandas, NumPy, and visualization tools such as Matplotlib and Power BI. Skilled in data cleaning, exploratory data analysis, predictive modeling, and communicating findings to support data-informed decision-making in business and research environments."
```

```python
# 1. Clean the resume text
cleaned_resume = text_clean(resume1)

# 2. Convert to numerical form using the fitted tfidf model
# This uses the process_text_to_vector function defined earlier
resume_vector = process_text_to_vector(resume1, tfidf)

# 3. Print results
print("Cleaned Text Sample:")
print(cleaned_resume[:100], "...")

print("\nNumerical Vector (Sparse Matrix Representation):")
print(resume_vector)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

# Calculate similarity between the resume vector and all job description vectors
# resume_vector is (1, 763) and tfidf_matrix is (50, 763)
similarity_scores = cosine_similarity(resume_vector, tfidf_matrix)

# Add the scores to our original dataframe for easy viewing
df['similarity_score'] = similarity_scores.flatten()

# Sort by highest similarity and display the top 5 matches
top_matches = df[['title', 'similarity_score']].sort_values(by='similarity_score', ascending=False)

print("Top 5 Job Matches for Resume 1:")
display(top_matches.head(5))
```
